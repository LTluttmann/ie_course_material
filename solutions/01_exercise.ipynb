```python
import spacy
from rich import print as rprint
from ie_course.utils import normalize_whitespace, sentence_tokenize

nlp = spacy.load('en_core_web_sm')
```

## 1. Text Preprocessing

In real-world IE, data is often "dirty" (excessive whitespace, inconsistent casing, etc.). The first step is always normalization.


```python
raw_text = "  Information Extraction    is a subfield of NLP.   \n\nIt focuses on extracting structured information  from unstructured text.  "

cleaned_text = normalize_whitespace(raw_text).lower()
sentences = sentence_tokenize(cleaned_text)

print(f"Cleaned: {cleaned_text}")
print(f"Sentences: {sentences}")
```

## 2. Simple Extraction Task

Extract the **Amount** and **Currency** from the email text.


```python
email_text = "Please find the attached invoice for the amount of 1,250.00 EUR. Payment is due by next Friday."

# Fixed extraction
amount = email_text.split("amount of ")[1].split(" ")[0].strip()
currency = email_text.split(str(amount))[1].split(".")[0].strip()
print(amount)
print(currency)
```

## 3. IR vs IE

Perform:
1. IR: retrieve snippets containing "Cupertino"
2. IE: extract organizations


```python
snippets = [
    "Apple Inc. announced a new iPhone in Cupertino.",
    "The weather in Cupertino is sunny today.",
    "Microsoft is expanding its offices in Redmond."
]

query = "Cupertino"

# IR
retrieved_docs = [doc for doc in snippets if query in doc]

# IE
extracted_orgs = []
for doc_text in retrieved_docs:
    doc = nlp(doc_text)
    for ent in doc.ents:
        if ent.label_ == "ORG":
            extracted_orgs.append(ent.text)

rprint(f"Retrieved Docs (IR): {retrieved_docs}")
rprint(f"Extracted Orgs (IE): {extracted_orgs}")
```
