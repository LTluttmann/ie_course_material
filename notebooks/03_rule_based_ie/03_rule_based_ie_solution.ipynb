## 2. Preprocessing OCR Text

OCR text often contains character confusions. However, blindly replacing characters can introduce new errors.

For example, replacing every `O` with `0` may fix `INV-2O24-OO7`, but it may damage real words such as `OCR`, `Support`, or company names.

In this task, start with a simple preprocessing function and improve it carefully.

````python
def preprocess_ocr_text(text: str) -> str:
    """Basic preprocessing for OCR-like invoice text."""
    text = re.sub(r"\s+", " ", text)
    return text.strip()


cleaned_text = preprocess_ocr_text(ocr_invoice_text)
print(cleaned_text)
````

## Task 2 — Improve Preprocessing

Improve the preprocessing function so it fixes some OCR errors without being too aggressive.

### Suggestions

- Fix `2O24` → `2024`
- Fix `OO7` → `007`
- Fix `0S.02.2024` → `05.02.2024`
- Fix numeric amounts such as `12O,00` and `135,OO`

### Important

Avoid replacing all `O` characters globally unless you can explain why this is risky.

````python
import re

def improved_preprocess_ocr_text(text: str) -> str:
    
    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)
    
    # --- Fix common OCR errors ONLY in numeric contexts ---
    
    # Replace 'O' with '0' when surrounded by digits (e.g., 12O -> 120)
    text = re.sub(r'(?<=\d)O(?=\d)', '0', text)
    
    # Replace 'O' at end of number (e.g., 135,OO -> 135,00)
    text = re.sub(r'(?<=\d)O\b', '0', text)
    
    # Replace 'OO' in numbers (e.g., OO7 -> 007)
    text = re.sub(r'\bO{2,}(?=\d)', lambda m: '0' * len(m.group()), text)
    
    # Replace 'S' with '5' in dates (e.g., 0S.02.2024 -> 05.02.2024)
    text = re.sub(r'(?<=\d)S(?=\.)', '5', text)
    
    # Replace 'O' with '0' in dates (e.g., 19.O2.2024 -> 19.02.2024)
    text = re.sub(r'(?<=\.)O(?=\d)', '0', text)
    
    # --- Optional: normalize currency spacing ---
    text = re.sub(r'\s+EUR', ' EUR', text)
    
    return text.strip()
````

## 3. Extract the Invoice Number

The invoice number appears after `Rechnung Nr.`.

Because the OCR output contains character errors, extraction may fail unless preprocessing handles them.

````python
def extract_invoice_number(text: str) -> str | None:
    pattern = r"Rechnung\s*Nr\.?\s*:?\s*([A-Z0-9-]+)"
    match = re.search(pattern, text, flags=re.IGNORECASE)
    return match.group(1) if match else None


invoice_number = extract_invoice_number(improved_text)
print(invoice_number)
````

## 4. Extract the Invoice Date

The invoice date appears after `Datum:`.

The OCR output contains `0S.02.2024`, where `S` should probably be `5`.

````python
def extract_date(text: str) -> str | None:
    pattern = r"Datum\s*:\s*(\d{2}\.\d{2}\.\d{4})"
    match = re.search(pattern, text, flags=re.IGNORECASE)
    return match.group(1) if match else None


date = extract_date(improved_text)
print(date)
````

## 5. Extract the Total Amount

The total amount appears after `Gesamtbetrag:`.

The amount uses a European number format:

```text
1.136,45 EUR
```

Here, the dot is a thousands separator and the comma is a decimal separator.

````python
def extract_total_amount(text: str) -> str | None:
    pattern = r"Gesamtbetrag\s*:\s*([\d\.,]+)\s*EUR"
    match = re.search(pattern, text, flags=re.IGNORECASE)
    return match.group(1) if match else None


amount = extract_total_amount(improved_text)
print(amount)
````

## 6. Normalize the Amount

Convert the extracted amount string into a Python `float`.

Example:

```text
1.136,45 → 1136.45
```

````python
def normalize_euro_amount(amount: str | None) -> float | None:
    if amount is None:
        return None
    
    amount = amount.replace(".", "")
    amount = amount.replace(",", ".")
    
    try:
        return float(amount)
    except ValueError:
        return None


normalized_amount = normalize_euro_amount(amount)
print(normalized_amount)
````

## 7. Build the Full Rule-Based IE Pipeline

Now combine all steps into one pipeline:

```text
OCR-like text → preprocessing → regex extraction → normalization → structured output
```

````python
def extract_invoice_fields(text: str) -> dict:
    cleaned = improved_preprocess_ocr_text(text)
    amount = extract_total_amount(cleaned)
    
    return {
        "invoice_number": extract_invoice_number(cleaned),
        "date": extract_date(cleaned),
        "total_amount": normalize_euro_amount(amount),
    }


result = extract_invoice_fields(ocr_invoice_text)
pprint(result)
````

````python
def improved_preprocess_ocr_text(text):
    text = re.sub(r"\s+", " ", text)

    # Fix OCR mistakes in numeric contexts
    text = re.sub(r"(?<=\d)O(?=\d)", "0", text)   # 2O24 -> 2024
    text = re.sub(r"(?<=\d)O\b", "0", text)       # 135,OO -> 135,O
    text = re.sub(r"(?<=,)O\b", "0", text)        # 135,O -> 135,0
    text = re.sub(r"\bO(?=\d)", "0", text)        # O3 -> 03

    # Fix repeated OCR O's in IDs/numbers
    text = re.sub(r"OO(?=\d)", "00", text)
    text = re.sub(r"(?<=-)OO(?=\d)", "00", text)

    # Fix S misread as 5 in dates
    text = re.sub(r"(?<=\d)S(?=\.)", "5", text)

    # Fix I/l confusion in common labels
    text = text.replace("TotaI", "Total")
    text = text.replace("Pr1ce", "Price")

    return text.strip()
````

````python
def extract_invoice_number(text):
    patterns = [
        r"Rechnung\s*Nr\.?\s*:?\s*([A-Z0-9-]+)",
        r"Invoice\s*N[o0]\.?\s*:?\s*([A-Z0-9-]+)",
        r"Invoice\s*Number\s*:?\s*([A-Z0-9-]+)",
        r"Inv\.?\s*ID\s*:?\s*([A-Z0-9-]+)",
    ]

    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)
        if match:
            return match.group(1)

    return None
````

````python
def extract_date(text):
    patterns = [
        r"Datum\s*:\s*(\d{2}\.\d{2}\.\d{4})",
        r"Date\s*:\s*(\d{4}-\d{2}-\d{2})",
        r"Date\s*:\s*(\d{2}\.\d{2}\.\d{4})",
        r"Date\s*:\s*(\d{2}/\d{2}/\d{4})",
    ]

    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)
        if match:
            return match.group(1)

    return None
````

````python
def extract_total_amount(text):
    patterns = [
        r"Gesamtbetrag\s*:\s*([\d\.,]+)\s*EUR",
        r"Total\s*Amount\s*Due\s*:\s*EUR\s*([\d\.,]+)",
        r"Total\s*Amount\s*Due\s*:\s*([\d\.,]+)\s*EUR",
        r"Total\s*Due\s*:\s*EUR\s*([\d\.,]+)",
        r"Total\s*:\s*EUR\s*([\d\.,]+)",
    ]

    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)
        if match:
            return match.group(1)

    return None
````

````python
def normalize_amount(amount: str | None) -> float | None:
    if amount is None:
        return None

    amount = amount.strip()

    # case: european format. e.g. 1.136,45
    if "." in amount and "," in amount:
        # If comma appears after dot, assume dot = thousands, comma = decimal
        if amount.rfind(",") > amount.rfind("."):
            amount = amount.replace(".", "")
            amount = amount.replace(",", ".")
        # otherwise assume comma = thousands, dot = decimal
        else:
            amount = amount.replace(",", "")

    # case: only comma. e.g. 89,99
    elif "," in amount:
        amount = amount.replace(",", ".")

    try:
        return float(amount)
    except ValueError:
        return None
````

````python
def extract_invoice_fields(text: str) -> dict:
    cleaned = improved_preprocess_ocr_text(text)
    amount = extract_total_amount(cleaned)

    return {
        "invoice_number": extract_invoice_number(cleaned),
        "date": extract_date(cleaned),
        "total_amount": normalize_amount(amount),
    }
````

## Reflection (sample answers)

1. **Which OCR errors were easiest to fix?**  
The easiest errors were simple character confusions inside numbers, like replacing 'O' with '0' in cases such as "2O24" or "12O,00". These are relatively safe because they appear in numeric contexts.

2. **Which OCR errors were dangerous to fix with simple replacement rules?**  
Global replacements like replacing all 'O' with '0' are dangerous, because they can break normal words like "OCR" or "Company". These kinds of fixes can introduce new errors instead of solving them.

3. **Which field was easiest to extract?**  
The invoice number was the easiest to extract, because it had a clear label ("Rechnung Nr." or "Invoice No") and a predictable format.

4. **Which field was hardest to extract?**  
The total amount was the hardest, because it can appear in different formats (EUR, €, commas, dots), may contain OCR errors, and requires additional normalization after extraction.

5. **Why do rule-based IE systems become brittle?**  
Rule-based systems become brittle because they rely on fixed patterns. When the input format changes (different labels, languages, or OCR errors), the rules often break and need manual updates.

6. **At what point would you consider using ML or LLM-based extraction instead?**  
When there are many different document formats, when rules become too complex to maintain, when OCR errors vary a lot, or when the system needs to generalize to new unseen data. In these cases, ML or LLM-based methods are more flexible and scalable.
