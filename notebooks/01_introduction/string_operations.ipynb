## Simplified information extration

```python
text = "Total: 120 USD"
parts = text.split(" ")

amount = parts[1]
currency = parts[2]

print("Amount: ", amount)
print("Currency: ", currency)
```

```python
address = """
Name: Steve Jobs
Street: 123 Business Park, London, UK
Zip Code: 62704
Country: United States
"""
```

```python
start_idx = address.lower().find("name:")+len("name:")
address[start_idx:].split("\n")[0].strip()
```

```python
# if we know that zip codes are 5 digits we can infer the end index
start_idx = address.lower().find("zip code: ")+len("zip code: ")
end_idx = start_idx + 5
address[start_idx:end_idx].strip()
```

```python
def extract_key_value(text, key_name):
    """
    In IE, we often look for 'Anchors' (the key) to find the 'Value'.
    This function mimics basic field extraction.
    """
    # Normalize key for search
    key_search = key_name.lower()
    
    if key_search in text:
        # Find the starting index of the value after the key and colon
        start_index = text.find(key_search) + len(key_search) + 1
        # Get the line containing the value
        line = text[start_index:].split('\n')[0]
        return line.strip()
    return None
```
